### AI, Machine Learning, and Data Science

People who work in *AI* are trying to put intelligence into a computer. In many cases, they're trying to emulate a brain inside of computer software and hardware.

*Machine Learning* is a subset of methods that came out of AI research that allows a machine to get better at learning the relationship between various concepts given some data.

*Deep Learning* is a specific algorithm that originally sought to mimic the brain. We know these models don't line up with actual brain function, for the most part, BUT they've been found to be incredibly useful for many tasks.

## Training a Computer Vision Model (Teachable Machine)

As a class, we are going to train a computer vision model from objects we have in the classroom using Teachable Machine ([https://teachablemachine.withgoogle.com/](https://teachablemachine.withgoogle.com/)).

If you want to come back to this site, visit the link and follow these steps:
1. Click the `Get Started` button, this will open a page to create a *New Project*.
1. On the *New Project* page, select the `Image Project` option.
1. There will be two options for an `Image Project`, the `Standard image model` option will work for this activity.
1. Using the new interface, create some appopriate names for the classes (think of them as labels) and use the webcam to take lots of pictures of the selected objects.
1. Once you feel comfortable with your data for the classes you created, click the `Train Model` button, this will create a model that we can *preview* on the webpage.



Right now, you all need to find an object that you want our classroom model to be able to recognize. Each person will bring an object up to this computer and we will use Teachable Machine to collect the data using the webcam.

As we're collecting data consider these questions:
1. How should the object be held?
1. Should it move around? How much?
1. What are your reasonings for your responses?



Once we've trained the model, let's test it with object we used.

**But first, what objects do you think our model will have the hardest time with?**

As we are testing the model, think about these questions:
1. Does our model work?
