## Inheritance

Often, we use classes when we want to create lots of very similar things, like different kinds of animals.

Classes can also be "sub-classes", or child classes, or derived classes from other classes like in the example below.

```python
class Animal:

    def __init__(self, number_of_legs, does_it_fly, mammal, claws, weight):
        self.number_of_legs = number_of_legs
        self.does_it_fly = does_it_fly
        self.mammal = mammal
        self.claws = claws
        self.weight = weight

    def eat(self, pounds):
        self.weight += pounds

    def sleep(self):
        self.weight -= 5

    def hunt(self):
        self.weight -= 5

class Tiger(Animal):

    def __init__(self, number_of_legs, does_it_fly, mammal, claws, weight, type, has_camo,):
        super().__init__(number_of_legs, does_it_fly, mammal, claws, weight)
        self.type = type

daniel = Tiger(4,False,True,True,150, "Bengal", True)
bob = Tiger(4,False, True, True, 200, "Siberian", True)

print(daniel.weight)
daniel.eat(20)
print(daniel.weight)
print(bob.weight)
bob.sleep()
print(bob.weight)
print(bob.type)
```

## Creating a class of your own animal

```python
class Sloth(Animal):

#aniimal of choice: sloth
    
    def __init__(self, number_of_legs, two_toed, three_toed, mammal, weight, fur, omnivore, type):
        self.number_of_legs = number_of_legs
        self.two_toed = two_toed
        self.three_toed = three_toed
        self.mammal = mammal
        self.weight = weight
        self.does_it_have_fur = fur
        self.is_it_an_omnivore = omnivore
        self.type = type

seth = Sloth(4, False, True, True, 9, True, True, "Maned Sloth")
gary = Sloth(4, True, False, True, 13, True, True, "Linnaeus Sloth")
dave = Sloth(4, False, True, True, 6, True, True, "Pygmy Sloth")

print(seth.type)
print(gary.type)
print(dave.type)
```
